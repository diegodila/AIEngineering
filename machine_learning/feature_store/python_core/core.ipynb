```python
class Moto:
    def __init__(self, marca: str, cilindrada: int, dados: list[str]) -> None:
        self.marca = marca
        self.cilindrada = cilindrada
        self.dados = dados


class Pessoa:
    def __init__(self, nome: str, moto: Moto) -> None:
        self.nome = nome
        self.moto = moto


p1 = Pessoa("diego", Moto("Harley", 1200, ["1","2","3"]))
```

```python
def gera_numeros():
    yield 1
    yield 2
    yield 3

# Usando o gerador com um loop for
for numero in gera_numeros():
    print(numero)
```

```python
gera_numeros()
```

```python
list(gera_numeros())
```

```python
def pegar_itens(caixa: list[list[str]]) -> Generator[Unknown, Unknown, None]:
    for item in caixa:
        if isinstance(item, list):
            # ❌ Sem 'from': entrega a caixa fechada
            # yield pegar_itens(item)

            # ✅ Com 'from': abre a caixa e entrega elemento por elemento
            #yield from pegar_itens(item)
            yield from pegar_itens(item)
        else:
            yield item


minha_caixa : list[str | list[str | list[str]]] = ["Livro", ["Caneta",["ds", "sd"], "Lápis"], "Caderno"]

print(list(pegar_itens(minha_caixa)))
# Saída: ['Livro', 'Caneta', 'Lápis', 'Caderno']
```
